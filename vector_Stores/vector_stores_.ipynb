```python
vector_Store = Chroma(
    collection_name="football_players",
    embedding_function=embeddings,
    persist_directory="./chroma_db")
```

```python
vector_Store.add_documents(docs)
```

```python
vector_Store.get(include=['embeddings','documents','metadatas'])
```

```python
answer = vector_Store.similarity_search(
    query="who is the best player in football",
    k=1
)
type(answer)
```

```python
# for i in answer:
#   print(i)
answer[0].page_content
```

```python
ans2 = vector_Store.similarity_search(
    query="who is the worst player in the football",
    k=1

)
for i in ans2:
  print(i.page_content)
```

```python
#answers with similarity scores
ans3 = vector_Store.similarity_search_with_score(
    query="give me the name of 3 best football players",
    k=3
)
for i in ans3:
  print(i[0].page_content ,"similarity Score:", i[1])
```

```python
#metadata based searching
ans4 = vector_Store.similarity_search_with_score(
    filter={"ballon_dor": 0},
    query="give me the name of 4 best football players",
    k=4
)
for i in ans4:
  print(i[0].page_content ,"similarity Score:", i[1])
```
